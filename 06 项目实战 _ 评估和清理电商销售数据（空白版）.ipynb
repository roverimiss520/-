# 项目：评估和清理英国电商公司销售数据

## 分析目标

此数据分析的目的是，根据市场销售数据，挖掘畅销产品，以便制定更有效的市场策略来提升营收。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了一家英国在线零售公司在2010年12月1日至2011年12月9日期间的所有交易情况，涵盖了该公司在全球不同国家和地区的业务数据。该公司主要销售覆盖各个场景的礼品，包括但不限于生日礼品、结婚纪念品、圣诞礼品等等。该公司的客户群体主要包括批发商和个人消费者，其中批发商占据了相当大的比例。

数据每列的含义如下：
- `InvoiceNo`: 发票号码。6位数，作为交易的唯一标识符。如果这个代码以字母“c”开头，表示这笔交易被取消。
- `StockCode`: 产品代码。5位数，作为产品的唯一标识符。
- `Description`: 产品名称。
- `Quantity`: 产品在交易中的数量。
- `InvoiceDate`: 发票日期和时间。交易发生的日期和时间。
- `UnitPrice`: 单价。价格单位为英镑（£）。
- `CustomerID`: 客户编号。5位数，作为客户的唯一标识符。
- `Country`: 国家名称。客户所居住的国家的名称。

# 清理数据

```python
df2 = df1
```

```python
df2
```

```python
df2.dropna(subset = "Description",inplace = True)
```

```python
df2["Description"].isnull().sum()
```

目前清理了商品名称缺失同时商品单价为零的数据，并且检查无误

接下来保留商品销售数量大于零的商品，商品单价大于等于零的商品

```python
df2 = df2[df2["Quantity"] > 0]
```

```python
(df2["Quantity"] < 0).sum()
```

```python
df2 = df2[df2["UnitPrice"] >= 0 ]
```

```python
(df2["UnitPrice"] < 0).sum()
```

目前已经清楚商品售卖数量、商品单价小于的数据，检查无误

接下来改变交易时间数据类型与客户id数据类型

```python
df2.info()
```

```python
df2["InvoiceDate"] = pd.to_datetime(df2["InvoiceDate"])
df2["InvoiceDate"].info()
```

```python
df2.info()
```

```python
df2["CustomerID"] = df2["CustomerID"].astype("str")
df2.info()
```

```python
df2[["CustomerID"]]
```

日期数据和顾客id数据类型已经改变，顾客id后面出现了".0",这些数据应该删除

使用slice函数清除

```python
df2["CustomerID"] = df2["CustomerID"].str.slice(0,-2)
df2[["CustomerID"]]
```

已经完成了".0"的处理

接下来处理国家的数据不一致

```python
df2["Country"].value_counts()
```

```python
df2["Country"] = df2["Country"].replace({"U.K.": "United Kingdom","UK":"United Kingdom","USA":"United States"} )
len(df2[df2["Country"] == "U.K."])
```

```python
len(df2[df2["Country"] == "UK"])
```

```python
len(df2[df2["Country"] == "USA"])
```

数据处理完成，进行保存
